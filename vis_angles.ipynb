```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load the CSV file
file_path = 'output_angles.csv'  # Replace with the actual file path
data = pd.read_csv(file_path, header=None)

# Plot each column on its own graph
for i in range(data.shape[1]):  # Iterate over columns
    plt.figure()  # Create a new figure for each column
    plt.plot(data.index, data[i], label=f'Column {i+1}')
    plt.title(f'Plot of Column {i+1}')
    plt.xlabel('Index')
    plt.ylabel('Value')
    plt.legend()
    plt.grid(True)
    plt.show()
```

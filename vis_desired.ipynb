```python
import matplotlib.pyplot as plt
```

```python
import csv

# Initialize lists to store each column
desired_0 = []
desired_1 = []
desired_2 = []

current_0 = []
current_1 = []
current_2 = []

# Read the CSV file
with open('positions.csv', 'r') as file:
    reader = csv.reader(file)
    # Skip the header row if it exists
    next(reader, None)
    for row in reader:
        desired_0.append(row[0])
        desired_1.append(row[1])
        desired_2.append(row[2])
        current_0.append(row[3])
        current_1.append(row[4])
        current_2.append(row[5])
```

```python
desired_0
```

```python
plt.plot(desired_0, label='desired_0', marker='o')
plt.plot(current_0, label='current_0', marker='x')
plt.xlabel('Index')
plt.ylabel('Value', labelpad=15)  # Adjust labelpad to add spacing
plt.title('Comparison of desired_0 and current_0')
plt.legend()
plt.grid()
plt.show()
```
